```python
import pandas as pd
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.preprocessing import StandardScaler
import numpy as np

# Standardized input values provided by test runner
wing = float(input())
weight = float(input())
culmen = float(input())

# Read data and prepare features
Hawks = pd.read_csv("hawks.csv").dropna()

# Extract target array y
y = Hawks["Species"]

# Standardize the input features and store them in X
scaler = StandardScaler()
X = scaler.fit_transform(Hawks[["Wing", "Weight", "Culmen"]])

# Initialize and fit LDA model
LDAmodel = LinearDiscriminantAnalysis(n_components=2)
LDAmodel.fit(X, y)

# Print intercepts and weights
print(LDAmodel.intercept_)
print(LDAmodel.coef_)

# Pass standardized inputs directly to predict()
sample = [[wing, weight, culmen]]
preds = LDAmodel.predict(sample)

# Print matching ZyBooks output format
print(f"Predicted species is  {preds}")
```
